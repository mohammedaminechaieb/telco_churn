```python
import os, sys
if os.path.basename(os.getcwd()) == 'notebooks': os.chdir('..')
sys.path.insert(0, os.getcwd())
from data.load import load_clean, feature_columns
from data.preprocessing import split
from data.training import train_all
df = load_clean()
print(df.shape, df['Churn'].mean())
df.head()
```

```python
X_train, X_test, y_train, y_test = split(df)
num_cols, cat_cols = feature_columns(df)
fitted, best = train_all(X_train, y_train, num_cols, cat_cols, progress=lambda i, n: print(i, n))
best
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score
for n, p in fitted.items():
    pr = p.predict_proba(X_test[num_cols + cat_cols])[:, 1]
    print(f'{n:20s} AUC={roc_auc_score(y_test, pr):.3f} PR-AUC={average_precision_score(y_test, pr):.3f}')
```
